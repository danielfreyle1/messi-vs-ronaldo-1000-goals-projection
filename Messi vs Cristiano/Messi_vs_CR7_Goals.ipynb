```python

import pandas as pd
import numpy as np
from datetime import datetime
import kagglehub
import os

# Download latest version
path = kagglehub.dataset_download("nourheshamneuer/messi-vs-ronaldo-stats-by-nourhesham")

print("Path to dataset files:", path)
```

```python
# Revisar los archivos dentro de la ruta descargada
archivos_descargados = os.listdir(path)
print("Archivos encontrados:", archivos_descargados)

# Construir la ruta completa al archivo Excel
ruta_excel = os.path.join(path, archivos_descargados[0])

# Cargar el archivo y mostrar los nombres de todas las pestañas
archivo_completo = pd.ExcelFile(ruta_excel)
print("Hojas del dataset:", archivo_completo.sheet_names)

# Leer la hoja específica 'By Calendar Year' para ver la acumulación de goles
df_calendario = pd.read_excel(ruta_excel, sheet_name="By Calendar Year")

# Imprimir información sobre las variables y tipos de datos
print("\nInformación del dataset:")
df_calendario.info()

# Imprimir las primeras y últimas filas para visualizar la estructura de los datos
print("\nPrimeros registros:")
print(df_calendario.head())

print("\nÚltimos registros:")
print(df_calendario.tail())
```
